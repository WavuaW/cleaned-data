```python
import pandas as pd
```

```python
data = pd.read_csv('./seniors.csv')
```

```python
# Display the first few rows to verify data was loaded correctly
print("Original data preview:")
print(data.head())
```

```python
print(f"Total entries: {len(data)}")
```

```python
# Define the counties we want to separate
counties = [
    "Prince George's County", 
    "Montgomery County", 
    "Anne Arundel County", 
    "Charles County", 
    "Fairfax County", 
    "Arlington", 
    "Washington DC"
]
```

```python
# Create a dictionary to store DataFrames for each county
county_dfs = {}
```

```python
# Separate the data by county
for county in counties:
    if county == "Washington DC":
        # Special case for DC since it might be stored differently
        county_dfs[county] = data[data['County'] == "Washington DC"]
    else:
        county_dfs[county] = data[data['County'] == county]
    
    print(f"{county}: {len(county_dfs[county])} entries")
```

```python
# Save each county to its own CSV file
for county, df in county_dfs.items():
    # Create a clean filename (replace spaces and apostrophes)
    filename = county.replace(" ", "_").replace("'", "").replace(",", "") + ".csv"
    
    # Save to CSV file
    df.to_csv(filename, index=False)
    print(f"Saved {len(df)} entries to {filename}")
```
