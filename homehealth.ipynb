```python
import pandas as pd
import os
```

```python
# Load the dataset
file_path = './data/marylandhomehealthcare_facilities - marylandhomehealthcare_facilities.csv'  # Update with your actual file path
home_health_data = pd.read_csv(file_path)
```

```python
print(f"Total number of agencies: {len(home_health_data)}")
```

```python
# Define towns/cities for each county
county_towns = {
    "Prince George's County, MD": [
        'UPPER MARLBORO', 'LANHAM', 'CLINTON', 'FORT WASHINGTON', 'HYATTSVILLE', 
        'KENSINGTON', 'BOWIE', 'COLLEGE PARK', 'GREENBELT', 'LAUREL', 'OXON HILL'
    ],
    "Montgomery County, MD": [
        'SILVER SPRING', 'ROCKVILLE', 'BETHESDA', 'GAITHERSBURG', 'OLNEY', 
        'GERMANTOWN', 'TAKOMA PARK', 'CHEVY CHASE', 'POTOMAC', 'WHEATON'
    ],
    "Anne Arundel County, MD": [
        'ANNAPOLIS', 'GLEN BURNIE', 'SEVERNA PARK', 'PASADENA', 'ODENTON', 
        'CROFTON', 'EDGEWATER', 'ARNOLD', 'MILLERSVILLE', 'LINTHICUM HEIGHTS'
    ],
    "Charles County, MD": [
        'LA PLATA', 'WALDORF', 'INDIAN HEAD', 'BRYANTOWN', 'PORT TOBACCO', 
        'HUGHESVILLE', 'WHITE PLAINS', 'NEWBURG', 'POMFRET', 'WELCOME'
    ]
}
```

```python
# Create directory for county data if it doesn't exist
output_dir = './healthcare_data'
os.makedirs(output_dir, exist_ok=True)
```

```python
# Process each county
for county_name, towns in county_towns.items():
    # Filter data for the current county's towns
    healthcare_data = home_health_data[home_health_data['City/Town'].str.upper().isin(towns)]
    
    # Add county name column
    healthcare_data = healthcare_data.copy()
    healthcare_data['County'] = county_name
    
    # Create filename (replace spaces and special characters)
    filename = county_name.replace(" ", "_").replace("'", "").replace(",", "") + '_home_health.csv'
    output_path = os.path.join(output_dir, filename)
    
    # Save to CSV
    healthcare_data.to_csv(output_path, index=False)
    print(f"Saved {len(healthcare_data)} agencies to {output_path}")
    
    # Display towns found in the data for this county
    unique_towns = healthcare_data['City/Town'].str.upper().unique()
    print(f"  Towns included: {', '.join(unique_towns)}")
```
