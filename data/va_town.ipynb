```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
    
pd.set_option('display.max_columns', None)
```

```python
# Load the CSV file",
file_path = './vahomehealthcare_facilities.csv'  # Update this path if your file is named differently\n",
df = pd.read_csv(file_path)
df.head()
```

```python
print(f"Dataset shape: {df.shape}")
```

```python
fairfax_cities = [
       'FAIRFAX', 'RESTON', 'VIENNA', 'MCLEAN', 'FALLS CHURCH', 
        'SPRINGFIELD', 'HERNDON', 'CENTREVILLE', 'CHANTILLY', 
        'BURKE', 'ANNANDALE', 'CLIFTON', 'OAKTON', 'GREAT FALLS',
        'LORTON', 'FAIRFAX STATION', 'MERRIFIELD', 'DUNN LORING',
        'TYSONS', 'TYSONS CORNER',
    ]
```

```python
arlington_cities = [
        'ARLINGTON',
    ]
```

```python
all_target_cities = fairfax_cities + arlington_cities
all_target_cities
```

```python
print(f"States in the dataset: {df['State'].unique()}")
```

```python
# Correct way to create va_df (without the comma)
va_df = df[df['State'] == 'VA']

```

```python
target_counties_df = va_df[va_df['City/Town'].isin(all_target_cities)]
target_counties_df
```

```python
# Optional: Save the filtered dataset to a new CSV file
target_counties_df.to_csv('./va_HomeHealthCare.csv', index=False)
```
