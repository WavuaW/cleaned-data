```python
import pandas as pd
import os
```

```python
# Read the CSV file",
file_path = './data/marylandHospice2_facilities.csv'
hospital_data = pd.read_csv(file_path)
```

```python
print(f"Total number of hospitals: {len(hospital_data)}")
print("Counties in the dataset:")
print(hospital_data['County/Parish'].value_counts())
```

```python
# Create a mapping from dataset county names to proper county names
county_mapping = {
    "Prince George's": "Prince George's County, MD",
    'Montgomery': "Montgomery County, MD",
    'Anne Arundel': "Anne Arundel County, MD",
    'Charles': "Charles County, MD"
}
```

```python
output_dir = './Hospiccee_data'
os.makedirs(output_dir, exist_ok=True)
```

```python
# Separate data by county and save to individual CSV files
for hospice_abbr, hospice_name in county_mapping.items():
    # Filter data for the current county
    Hospiccee_data = hospital_data[hospital_data['County/Parish'] == hospice_abbr]
    
    # Create filename (replace spaces and special characters)
    filename = hospice_name.replace(" ", "_").replace("'", "").replace(",", "") + '.csv'
    output_path = os.path.join(output_dir, filename)
    
    # Save to CSV
    Hospiccee_data.to_csv(output_path, index=False)
    print(f"Saved {len(Hospiccee_data)} hospitals to {output_path}")
```
