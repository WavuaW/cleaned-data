```python
import pandas as pd  # Primary data handling
import numpy as np   # Numerical operations
import matplotlib.pyplot as plt  # Basic visualization
import seaborn as sns  # Advanced visualization
```

```python
df = pd.read_csv('./data/marylandHospice2_facilities.csv')  # Replace with your file path

```

```python
# Display the first few rows to understand the data
print("Original dataset shape:", df.shape)

```

```python
# Check all locations in the dataset
print("\nUnique locations in the dataset:")
print(df['County/Parish'].value_counts())
```

```python
# Remove duplicates based on 'Facility Name' column
df_unique_facilities = df.drop_duplicates(subset=['Facility Name'])
df_unique_facilities
```

```python
# Filter for facilities in Maryland counties
maryland_df = df[
    (df['County/Parish'] == "Prince George's") | 
    (df['County/Parish'] == 'Montgomery') |
    (df['County/Parish'] == 'Anne Arundel') |
    (df['County/Parish'] == 'Charles')
]
```

```python
# If you want unique facilities in those locations:
filtered_unique_df = maryland_df.drop_duplicates(subset=['Facility Name'])
```

```python
# Display the results
print(filtered_unique_df)
```

```python
# Optional: Save the filtered dataset to a new CSV file
filtered_unique_df.to_csv('./data/marylandHospice2_facilities.csv', index=False)
```
